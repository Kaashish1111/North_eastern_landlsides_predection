```python
import pandas as pd
```

```python
INPUT_CSV = "ner_all_edges.csv"
OUTPUT_CSV = "guwahati_shillong_edges.csv"
```

```python
LAT_MIN, LAT_MAX = 25.45, 26.25
LON_MIN, LON_MAX = 91.60, 92.00
```

```python
def extract_corridor():
    df = pd.read_csv(INPUT_CSV, low_memory=False)
    print(f"Total edges in full NER network: {len(df)}")
 
    bbox_mask = (
        df["start_lat"].between(LAT_MIN, LAT_MAX)
        & df["start_lon"].between(LON_MIN, LON_MAX)
        & df["end_lat"].between(LAT_MIN, LAT_MAX)
        & df["end_lon"].between(LON_MIN, LON_MAX)
    )
 
    nh6_mask = df["ref"].astype(str).str.contains("NH6", na=False)
 
    corridor_df = df[bbox_mask & nh6_mask].reset_index(drop=True)
    print(f"Edges after filtering to Guwahati-Shillong (NH6) corridor: {len(corridor_df)}")
 
    corridor_df.to_csv(OUTPUT_CSV, index=False)
    print(f"Saved to {OUTPUT_CSV}")
    return corridor_df
 
 
if __name__ == "__main__":
    extract_corridor()
```
